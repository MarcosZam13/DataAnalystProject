```python
import pandas as pd
import glob
import re

ruta = "/content/drive/MyDrive/ProyectoAnálisis/*.csv"
files = glob.glob(ruta)

dataframes = []

for file in files:
    df = pd.read_csv(file)

    # extraer año del nombre del archivo
    year = re.search(r'\d{4}', file).group()
    df["year"] = int(year)

    dataframes.append(df)

print(f"Se cargaron {len(dataframes)} datasets.")
```

```python
#Busqueda de columnas diferentes
todas_columnas = set()

for df in dataframes:
    todas_columnas.update(df.columns)

print("Total columnas diferentes encontradas:", len(todas_columnas))

for col in sorted(todas_columnas):
    print(col)
```

```python
for df in dataframes:
    print("year:", df["year"].iloc[0])
    print(len(df.columns), "columnas")
    print(df.columns.tolist())
    print()
```

```python
for i in range(len(dataframes)):
  # Check if the 'Year' column exists in the current dataframe
  if "Year" in dataframes[i].columns:
    # If it exists, drop it and update the dataframe in the list
    dataframes[i] = dataframes[i].drop(columns=["Year"])
```

```python
#Renombrado de columnas
rename_dict = {

# nombres
"short_name": "name",
"Name": "name",

# posición
"player_positions": "position",
"Position": "position",

# overall
"Overall": "overall",

# club
"club_name": "club",
"Club": "club",

# nación
"nationality_name": "nation",
"Nation": "nation",

# físico
"physic": "physicality",
"Physicality": "physicality",

# edad
"Age": "age",

# pace
"Pace": "pace",

# shooting
"Shooting": "shooting",

# passing
"Passing": "passing",

# dribbling
"Dribbling": "dribbling",

# defending
"Defending": "defending",

# weak foot
"Weak foot": "weak_foot",

# preferred foot
"Preferred foot": "preferred_foot",

# skill moves
"Skill moves": "skill_moves",

# aceleración
"Acceleration": "movement_acceleration",

# sprint
"Sprint": "movement_sprint_speed",

# stamina
"Stamina": "power_stamina",

# year
"Year": "year"
}

for i in range(len(dataframes)):
    dataframes[i] = dataframes[i].rename(columns=rename_dict)
```

```python
#Selección de columnas finales para cada .csv
columnas_finales = [
"name",
"position",
"overall",
"potential",
"value_eur",
"wage_eur",
"age",
"height_cm",
"weight_kg",
"league_name",
"league_level",
"club",
"nation",
"preferred_foot",
"weak_foot",
"skill_moves",
"pace",
"shooting",
"passing",
"dribbling",
"defending",
"physicality",
"movement_acceleration",
"movement_sprint_speed",
"power_stamina",
"year"
]

for df in dataframes:

    y = df["year"].iloc[0]

    faltantes = [c for c in columnas_finales if c not in df.columns]

    if faltantes:
        print(f"Año {y} le faltan columnas:")
        print(faltantes)
        print()
```

```python
#Colocar las columnas faltantes y rellenarlo de nulos

for i in range(len(dataframes)):

    for col in columnas_finales:
        if col not in dataframes[i].columns:
            dataframes[i][col] = pd.NA
```

```python
#Seleccionar solo las columnas finales para cada .csv
for i in range(len(dataframes)):
    dataframes[i] = dataframes[i][columnas_finales]
```

```python
#Unir los 9 csv
df_total = pd.concat(dataframes, ignore_index=True)

print("Shape del dataset final:", df_total.shape)

df_total.info()

df_total.head()
```

```python
#Tipado

# columnas categóricas
categoricas = [
"name",
"position",
"league_name",
"club",
"nation",
"preferred_foot"
]

# columnas enteras
enteras = [
"overall",
"potential",
"age",
"height_cm",
"weight_kg",
"weak_foot",
"skill_moves",
"pace",
"shooting",
"passing",
"dribbling",
"defending",
"physicality",
"movement_acceleration",
"movement_sprint_speed",
"power_stamina",
"league_level",
"year"
]

# columnas float
floats = [
"value_eur",
"wage_eur"
]

# aplicar tipado
df_total[categoricas] = df_total[categoricas].astype("category")
df_total[enteras] = df_total[enteras].astype("Int64")
df_total[floats] = df_total[floats].astype("float")

df_total.dtypes
```

```python
#Busqueda de nulos
nulos = df_total.isnull().sum()
print(nulos)

porcentaje_nulos = (df_total.isnull().sum() / len(df_total)) * 100

print(porcentaje_nulos.sort_values(ascending=False))
```

```python
#Reporte de nulos
reporte_nulos = pd.DataFrame({
    "columna": df_total.columns,
    "nulos": df_total.isnull().sum(),
    "porcentaje": (df_total.isnull().sum() / len(df_total)) * 100
})

reporte_nulos = reporte_nulos.sort_values("porcentaje", ascending=False)

reporte_nulos
```

```python
#Nulos de peso y altura se busca al jugador en otros años y se copia la misma
df_total = df_total.sort_values(["name", "year"])

df_total["height_cm"] = df_total.groupby("name")["height_cm"].transform(lambda x: x.ffill().bfill())
df_total["weight_kg"] = df_total.groupby("name")["weight_kg"].transform(lambda x: x.ffill().bfill())
```

```python
#Precio mercado y valor del jugador se saca con la mediana
df_total["value_eur"] = df_total["value_eur"].fillna(df_total["value_eur"].median())
df_total["wage_eur"] = df_total["wage_eur"].fillna(df_total["wage_eur"].median())
```

```python
#Busqueda de los nulos por posición:
stats = [
"pace",
"shooting",
"passing",
"dribbling",
"defending",
"physicality"
]

df_total[df_total[stats].isnull().any(axis=1)]["position"].value_counts()
```

```python
#Se analizó que los nulos eran de porteros, entonces se les coloca 0
df_total.loc[df_total["position"] == "GK", stats] = 0
```

```python
#Se imputa el potencial con respecto al overall
df_total["potential"] = df_total.groupby("overall")["potential"].transform(
    lambda x: x.fillna(x.median())
)
```

```python
#Cada club pertenece a una liga entonces se hace un busqueda para rellenar el nombre de la liga:
club_league_map = df_total.groupby("club")["league_name"].agg(lambda x: x.mode().iloc[0] if not x.mode().empty else None)

df_total["league_name"] = df_total["league_name"].fillna(df_total["club"].map(club_league_map))
```

```python
#Misma lógica para el league_level
club_level_map = df_total.groupby("club")["league_level"].agg(lambda x: x.mode().iloc[0] if not x.mode().empty else None)

df_total["league_level"] = df_total["league_level"].fillna(df_total["club"].map(club_level_map))
```

```python
nulos = df_total.isnull().sum()
print(nulos)

(df_total.isnull().sum() / len(df_total)) * 100
```

```python
#A los jugadores que no se les encontró club se le coloca como agentes libres
df_total["club"] = df_total["club"].cat.add_categories("Free Agent")
df_total["club"] = df_total["club"].fillna("Free Agent")
```

```python
#Se vuelve a mapear el nombre de la liga y el nivel
df_total["league_name"] = df_total["league_name"].fillna(df_total["club"].map(club_league_map))
df_total["league_level"] = df_total["league_level"].fillna(df_total["club"].map(club_level_map))
```

```python
#Los jugadores que no se encontraron en años anteriores se les rellena la altura y peso con la mediana
df_total["height_cm"] = df_total["height_cm"].fillna(df_total["height_cm"].median())
df_total["weight_kg"] = df_total["weight_kg"].fillna(df_total["weight_kg"].median())
```

```python
#Las ligas que no se encontraron se les coloca Unknown
df_total["league_name"] = df_total["league_name"].cat.add_categories("Unknown")
df_total["league_name"] = df_total["league_name"].fillna("Unknown")
```

```python
#El nivel de la liga se llena con la moda
df_total["league_level"] = df_total["league_level"].fillna(
    df_total["league_level"].mode()[0]
)
```

```python
nulos = df_total.isnull().sum()
print(nulos)

(df_total.isnull().sum() / len(df_total)) * 100
```

```python
#Busqueda de duplicados
duplicados = df_total.duplicated().sum()

print("Filas duplicadas:", duplicados)

porcentaje_dup = (duplicados / len(df_total)) * 100

print("Porcentaje duplicados:", porcentaje_dup, "%")
```

```python
df_total[df_total.duplicated(
    subset=["name","age","club","position","year"],
    keep=False
)].sort_values(["name","year"])
```

```python
#Se eliminan duplicados por nombre, edad, club, position y año
df_total = df_total.drop_duplicates(
    subset=["name","age","club","position","year"],
    keep="first"
)
```

Preparación y Limpieza del Dataset
1. Selección de columnas relevantes

A partir de los datasets originales de FIFA (2015–2024), se realizó una selección de variables relevantes para el análisis. Se eliminaron columnas redundantes o poco útiles para el objetivo del proyecto, manteniendo únicamente atributos relacionados con:

Identificación del jugador

Características físicas

Habilidades técnicas

Información económica

Contexto competitivo (liga, club)

Año de registro

Las columnas finales seleccionadas fueron:

name, position, overall, potential, value_eur, wage_eur, age,
height_cm, weight_kg, league_name, league_level, club, nation,
preferred_foot, weak_foot, skill_moves, pace, shooting, passing,
dribbling, defending, physicality, movement_acceleration,
movement_sprint_speed, power_stamina, year

Esta selección permite mantener un dataset compacto pero informativo, adecuado para análisis estadístico y generación de variables derivadas.

2. Integración de datasets

Se cargaron los 9 archivos CSV correspondientes a las ediciones de FIFA entre 2015 y 2024 utilizando pandas y glob.

Durante este proceso:

Se añadió una columna year para identificar la temporada correspondiente.

Se estandarizaron los nombres de columnas para mantener consistencia entre los datasets.

Finalmente, todos los datasets fueron concatenados en una sola tabla unificada.

Esto permitió construir un dataset longitudinal de jugadores a través de múltiples temporadas.

5. Limpieza e imputación de valores nulos

Se aplicaron diferentes estrategias de imputación dependiendo del tipo de variable y su significado dentro del dataset.

Variables físicas

Las columnas height_cm y weight_kg se imputaron inicialmente utilizando información histórica del mismo jugador en diferentes años.
Cuando no existía información disponible, se utilizó la mediana global de la variable para completar los valores faltantes.

Variables económicas

Las variables value_eur y wage_eur presentan una distribución con alta presencia de outliers, por lo que se utilizó la mediana como método de imputación para evitar sesgos.

Potencial del jugador

Los valores faltantes de potential se imputaron utilizando la mediana dentro de grupos de overall, debido a la fuerte relación existente entre ambas variables en el sistema de rating de FIFA.

Estadísticas de juego

Las columnas:

pace

shooting

passing

dribbling

defending

physicality

presentaban valores nulos principalmente en jugadores porteros (GK).
En estos casos se asignó el valor 0, ya que dichas estadísticas no aplican directamente a la posición de portero.

Información de liga

Para las variables:

league_name

league_level

se utilizó información del club del jugador para inferir los valores faltantes.
En los casos donde no fue posible inferir la liga, se asignó:

"Unknown" para league_name

la moda para league_level

6. Eliminación de duplicados

Debido a la integración de múltiples datasets, se detectaron registros duplicados.

Como el dataset no contiene un identificador único de jugador, se utilizó una clave compuesta para identificar duplicados:

(name, age, club, position, year)

Los registros duplicados fueron eliminados conservando la primera aparición de cada combinación.

7. Resultado final del proceso de limpieza

Después del proceso de limpieza:

Se eliminaron duplicados

Se imputaron valores faltantes

Se estandarizó el tipado de variables

El dataset final presenta:

0% de valores nulos en la mayoría de las variables

un porcentaje mínimo imputado en variables físicas y de liga

consistencia estructural entre todas las temporadas

Esto garantiza un dataset limpio, consistente y preparado para la etapa de análisis y generación de features derivados.
