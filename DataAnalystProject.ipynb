```python
import pandas as pd
import glob
import re

ruta = "/content/drive/MyDrive/ProyectoAnálisis/*.csv"
files = glob.glob(ruta)

dataframes = []

for file in files:
    df = pd.read_csv(file)

    # extraer año del nombre del archivo
    year = re.search(r'\d{4}', file).group()
    df["year"] = int(year)

    dataframes.append(df)

print(f"Se cargaron {len(dataframes)} datasets.")
```

```python
#Busqueda de columnas diferentes
todas_columnas = set()

for df in dataframes:
    todas_columnas.update(df.columns)

print("Total columnas diferentes encontradas:", len(todas_columnas))

for col in sorted(todas_columnas):
    print(col)
```

```python
for df in dataframes:
    print("year:", df["year"].iloc[0])
    print(len(df.columns), "columnas")
    print(df.columns.tolist())
    print()
```

```python
#Renombrado de columnas
rename_dict = {

# nombres
"short_name": "name",
"Name": "name",

# posición
"player_positions": "position",
"Position": "position",

# overall
"Overall": "overall",

# club
"club_name": "club",
"Club": "club",

# nación
"nationality_name": "nation",
"Nation": "nation",

# físico
"physic": "physicality",
"Physicality": "physicality",

# edad
"Age": "age",

# pace
"Pace": "pace",

# shooting
"Shooting": "shooting",

# passing
"Passing": "passing",

# dribbling
"Dribbling": "dribbling",

# defending
"Defending": "defending",

# weak foot
"Weak foot": "weak_foot",

# preferred foot
"Preferred foot": "preferred_foot",

# skill moves
"Skill moves": "skill_moves",

# aceleración
"Acceleration": "movement_acceleration",

# sprint
"Sprint": "movement_sprint_speed",

# stamina
"Stamina": "power_stamina",

# year
"Year": "year"
}

for i in range(len(dataframes)):
    dataframes[i] = dataframes[i].rename(columns=rename_dict)
```

```python
#Selección de columnas finales para cada .csv
columnas_finales = [
"name",
"position",
"overall",
"potential",
"value_eur",
"wage_eur",
"age",
"height_cm",
"weight_kg",
"league_name",
"league_level",
"club",
"nation",
"preferred_foot",
"weak_foot",
"skill_moves",
"pace",
"shooting",
"passing",
"dribbling",
"defending",
"physicality",
"movement_acceleration",
"movement_sprint_speed",
"power_stamina",
"year"
]

for df in dataframes:

    y = df["year"].iloc[0]

    faltantes = [c for c in columnas_finales if c not in df.columns]

    if faltantes:
        print(f"Año {y} le faltan columnas:")
        print(faltantes)
        print()
```

```python
#Colocar las columnas faltantes y rellenarlo de nulos

for i in range(len(dataframes)):

    for col in columnas_finales:
        if col not in dataframes[i].columns:
            dataframes[i][col] = pd.NA
```

```python
#Seleccionar solo las columnas finales para cada .csv
for i in range(len(dataframes)):
    dataframes[i] = dataframes[i][columnas_finales]
```

```python
#Unir los 9 csv
df_total = pd.concat(dataframes, ignore_index=True)

print("Shape del dataset final:", df_total.shape)

df_total.info()

df_total.head()
```

```python
#Busqueda de nulos
nulos = df_total.isnull().sum()
print(nulos)

porcentaje_nulos = (df_total.isnull().sum() / len(df_total)) * 100

print(porcentaje_nulos.sort_values(ascending=False))
```

```python
#Reporte de nulos
reporte_nulos = pd.DataFrame({
    "columna": df_total.columns,
    "nulos": df_total.isnull().sum(),
    "porcentaje": (df_total.isnull().sum() / len(df_total)) * 100
})

reporte_nulos = reporte_nulos.sort_values("porcentaje", ascending=False)

reporte_nulos
```

```python
#Busqueda de duplicados
duplicados = df_total.duplicated().sum()

print("Filas duplicadas:", duplicados)

porcentaje_dup = (duplicados / len(df_total)) * 100

print("Porcentaje duplicados:", porcentaje_dup, "%")
```
