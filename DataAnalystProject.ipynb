```python
import pandas as pd
import glob
import re

ruta = "/content/drive/MyDrive/ProyectoAnálisis/*.csv"
files = glob.glob(ruta)

dataframes = []

for file in files:
    df = pd.read_csv(file)

    # extraer año del nombre del archivo
    year = re.search(r'\d{4}', file).group()
    df["year"] = int(year)

    dataframes.append(df)

print(f"Se cargaron {len(dataframes)} datasets.")
```

```python
#Busqueda de columnas diferentes
todas_columnas = set()

for df in dataframes:
    todas_columnas.update(df.columns)

print("Total columnas diferentes encontradas:", len(todas_columnas))

for col in sorted(todas_columnas):
    print(col)
```
